## Inline Question 2:

AdaGrad, like Adam, is a per-parameter optimization method that uses the following update rule:

```
cache += dw**2
w += - learning_rate * dw / (np.sqrt(cache) + eps)
```

John notices that when he was training a network with AdaGrad that the updates became very small, and that his network was learning slowly. Using your knowledge of the AdaGrad update rule, why do you think the updates would become very small? Would Adam have the same issue?


## Answer:
[AdaGrad 容易因为历史梯度平方不断累积，导致有效学习率越来越小；Adam 通过指数移动平均缓解了这一问题，但并不能保证完全避免更新过小。]

AdaGrad 在训练过程中不断累积历史梯度的平方和，并利用该累积值调整每个参数的有效学习率。由于梯度平方始终非负，累积值会随迭代次数增加而不断增大，导致更新公式中的分母变大、有效学习率持续减小。因此，即使模型尚未收敛，参数更新也可能变得非常微小，使训练速度显著下降。

Adam 在一定程度上缓解了这一问题。与 AdaGrad 不同，Adam 使用梯度平方的指数移动平均，而不是无限累积历史梯度，因此能够减弱早期梯度的影响，并根据近期梯度动态调整更新步长。同时，Adam 结合一阶矩估计和偏差修正，通常能够保持更有效的参数更新。不过，Adam 仍可能因梯度过小或其他优化因素而出现更新缓慢的情况，并不能完全避免这一问题。
